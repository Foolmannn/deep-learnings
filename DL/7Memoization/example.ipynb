Without Memoization 

```python
def fibonacci(n):

    if n <= 1:
        return n

    return fibonacci(n-1) + fibonacci(n-2)


print(fibonacci(10))
```

```python
import time
start = time.time()
fibonacci(40)
print(time.time()-start)
```

With memoization 

```python
def fibonacci_mem(n, cache={}): # storing the calculated fibonacci in dictionary 

    if n in cache:
        return cache[n]

    if n <= 1:
        return n

    result = fibonacci_mem(n-1, cache) + fibonacci_mem(n-2, cache)

    cache[n] = result

    return result


print(fibonacci_mem(6))
```

```python
import time
start = time.time()
fibonacci_mem(40)
print(time.time()-start)
```
