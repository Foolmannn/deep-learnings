```python
import os
from PIL import Image
import pandas as pd

def organize_and_preprocess_dataset(source_dir, output_dir, label_mapping, train_split=0.8):
    """
    source_dir: Path to raw PNGs
    output_dir: Path where structured dataset will be created
    label_mapping: dict or DataFrame mapping filename to class label (e.g., {'34271.png': 'ka'})
    train_split: Percentage of data to go to training (e.g., 0.8 for 80%)
    """
    os.makedirs(os.path.join(output_dir, 'train'), exist_ok=True)
    os.makedirs(os.path.join(output_dir, 'test'), exist_ok=True)
    
    for filename, label in label_mapping.items():
        src_path = os.path.join(source_dir, filename)
        if not os.path.exists(src_path):
            continue
            
        # Determine whether it goes to train or test
        # (For a proper split, you can also randomize this or use a train_test_split from sklearn)
        subset = 'train' if (hash(filename) % 100) < (train_split * 100) else 'test'
        
        # Create class directory inside train/test
        class_dir = os.path.join(output_dir, subset, str(label))
        os.makedirs(class_dir, exist_ok=True)
        
        # Process image (Grayscale, 32x32, keep white background)
        with Image.open(src_path) as img:
            img = img.convert('L')
            if img.size != (32, 32):
                img = img.resize((32, 32), Image.Resampling.LANCZOS)
                
            # Save to the new structured path
            dest_path = os.path.join(class_dir, filename)
            img.save(dest_path)

    print("Dataset successfully structured and preprocessed!")
```
