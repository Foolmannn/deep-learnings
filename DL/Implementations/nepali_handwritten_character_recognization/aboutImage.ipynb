```python
from PIL import Image
import os

folder_path = "./data/dataset_200"
for filename in os.listdir(folder_path):
    if filename.endswith(".png"):
        img_path = os.path.join(folder_path, filename)
        with Image.open(img_path) as img:
            print(f"{filename}: Size (Width, Height) = {img.size}, Mode = {img.mode}")
        break  # Check the first image as a sample
```

```python

import numpy as np

def prepare_nepali_image(image_path):
    with Image.open(image_path) as img:
        # 1. Convert RGB to Grayscale (1 channel)
        img = img.convert('L')
        
        # 2. Ensure dimensions are strictly 32x32 (resize if necessary)
        if img.size != (32, 32):
            img = img.resize((32, 32), Image.Resampling.LANCZOS)
            
        # 3. Convert to numpy array and normalize to [0.0, 1.0]
        img_array = np.array(img, dtype=np.float32) / 255.0
        
        # Resulting shape will be (32, 32)
        return img_array
```

```python
X = prepare_nepali_image('./data/dataset_200/1.png')
```
