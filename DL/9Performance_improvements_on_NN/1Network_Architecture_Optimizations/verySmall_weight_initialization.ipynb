```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../data/ushape.csv')
```

```python
df.head()
```

```python
plt.scatter(df['X'],df['Y'],c=df['class'])
```

```python
X = df.iloc[:,0:2].values
y = df.iloc[:,-1].values
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(2,activation='sigmoid',input_dim=2))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
# Set parameters to 0
model.get_weights()
```

```python
initial_weights = model.get_weights()
```

Overwriting weights to new non zero constant value 

```python
initial_weights[0] = np.ones(model.get_weights()[0].shape)*0.5
initial_weights[1] = np.ones(model.get_weights()[1].shape)*0.5
initial_weights[2] = np.ones(model.get_weights()[2].shape)*0.5
initial_weights[3] = np.ones(model.get_weights()[3].shape)*0.5
```

```python
model.set_weights(initial_weights)
```

```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(X,y,epochs=100,validation_split=0.2)
```

```python
model.get_weights()
```

Although two output is coming from the single node the weights are same ie they are tied up as seen above . So model lost the non linearity , it acts as linear model 

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'), clf=model, legend=2)
```

We can see that the NN model is creating a line like linear model
